```python
import pandas as pd
import numpy as np
```

```python
from sklearn import preprocessing #for data preprocessing
```

```python
# create the data set:
data = pd.DataFrame({
    'age': [15,16,17,18,19,20],
    'studytime': [1,2,3,2,4,3],
    'failure': [0,1,0,2,1,0],
}).astype(float)
```

```python
#2. Min-Max Normalization:
# convert the values into ranges [0,1]
minmax_scaler = preprocessing.MinMaxScaler()
minmax_scaled = preprocessing.MinMaxScaler().fit_transform(data)
```

```python
#3. Z-Score Normalization
#mean = 0 and SD = 1
zscore_scaled = preprocessing.scale(data)
```

```python
#4.Decimal Scaling:
j = np.ceil(np.log10(data.abs().max()))
decimal_scaled = data / (10 ** j)
```

```python
#5. Display the Values:
print("Original Data:")
print(data)
```

```python
print("\n==== MIN-MAX NORMALIZATION =====")
print(pd.DataFrame(minmax_scaled,columns=data.columns))
```

```python
print("\n==== Z-SCORE NORMALIZATION =====")
print(pd.DataFrame(zscore_scaled,columns=data.columns))
```

```python
print("\n==== DECIMAL SCALLING NORMALIZATION =====")
print(decimal_scaled)
```
